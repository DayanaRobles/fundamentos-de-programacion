### Extra 1: Control de aforo con break y continue (while)
**Enunciado:** Crea un programa que procese visitantes del museo con un ciclo `while`. Por cada visitante se captura el costo de su boleto ($0 para menores de 3 años, $30 para 3 a 17 años, $45 para mayores de 18). Si el boleto es gratis, muestra "Menor de 3 años, sin cargo" y continúa con el siguiente visitante (usa `continue`). Si el total acumulado alcanza o supera $100, detén el ciclo con `break` y muestra el total.
**Entrada:**
```
Costo del boleto 1: 45
Costo del boleto 2: 0
Costo del boleto 3: 30
Costo del boleto 4: 45
```
**Salida:**
```
Menor de 3 años, sin cargo
Aforo alcanzado, se detiene el registro
Total acumulado: $120.00

````python
#Variables de acumulación que cambiaran según el registro
total = 0
#Numero de persona 
contador = 1
while True:
    #Solicitar al usuario que ingrese el costo del boleto
    #Guardarlas
    costo = float(input(f"Costo del boleto {contador}: "))
    #Si es costo es cero, imprimir que es sin cargo
    if costo == 0:
        contador += 1
        print("Menor de 3 años, sin cargo")
        continue

    else:
        contador += 1
        total += costo
        
    #Detenerse si se pasa de $100
    if total >= 100:
        print("Aforo alcanzado, se detiene el registro")
        break
#Mostrar el total acumulado
print(f"Total acumulado: ${total:.2f}")

    
````

### Extra 2: Estadística de visitantes con for
**Enunciado:** Crea un programa que pida el número total de visitantes y, con un ciclo `for`, capture la edad de cada uno. Debe contar cuántos son adultos (18 años o más) y calcular el promedio de edad de todos los visitantes, mostrando ambos resultados con dos decimales.
**Entrada:**
```
Número de visitantes: 4
Edad del visitante 1: 30
Edad del visitante 2: 45
Edad del visitante 3: 15
Edad del visitante 4: 2
```
**Salida:**
```
Adultos: 2
Promedio de edad: 23.00
```

````python
#Pedir número de visitantes:
personas = int(input("Número de visitantes: "))
#Variables de acumulación que cambiaran según el registro
edad_acumulada = 0
cont_mayores=0

for p in range(1, personas + 1):
    edad_persona = int(input(f"Edad del visitante {p}: "))
    #Si es mayor de edad, entra en el conteo además de acumular su edad
    if edad_persona >= 18:
        cont_mayores += 1
        edad_acumulada += edad_persona
    #Sino solo se acumula su edad
    else: 
        edad_acumulada += edad_persona
#Después del registro de personas, operación de promedio de edades
promedio_edades = edad_acumulada / personas
#Mostrar conteo de adultos
print(f"Adultos: {cont_mayores}")
#Mostrar Promedio de edad
print(f"Promedio de edad: {promedio_edades:.2f}")

````
