## 1. Descripción del reto

Desarrollar un programa en **Python** para cobrar las entradas de los visitantes que desean recorrer el **Museo de Antropología e Historia**, calculando el precio adecuado para cada visitante y aplicando descuentos por tipo de visitante bajo estrictas condiciones lógicas. El programa debe procesar a los visitantes mediante un **ciclo controlado**, aplicar los descuentos con una **tabla de verdad** que garantice un único descuento por boleto y desplegar el **total detallado** de todas las personas ingresadas.

**Precios base de entrada:**
| Tipo de visitante | Precio |
| :--- | :---: |
| Niños menores de 3 años | Gratis ($0) |
| Menores de edad (de 3 a 17 años) | $30 |
| Mayores de 18 años | $45 |

**Tabla de descuentos oficial:**
| Tipo de visitante | Descuento |
| :--- | :---: |
| Adulto mayor | 12% |
| Profesor | 10% |
| Estudiante | 10% |

Este reto integra las **estructuras de decisión** `if/elif/else` y los operadores lógicos (Tema 5), el **ciclo `while`** con `break` y `continue` (Tema 6), la **depuración con PDB** para garantizar un código libre de errores (Tema 7) y el **ciclo `for`** con acumuladores y procesamiento repetitivo (Tema 8).

```python
total=0
num_visitantes=int(input("Ingrese el número total de visitantes: "))

for v in range(1 , num_visitantes + 1):
    print(f" Hola! visitante {v} de {num_visitantes}")
    #While para checar si es una edad válida, no deja avanzar hasta que no ingrese una edad válida.
    #Beak rompe el while y continua
    while True:
        edad=int(input("Ingrese su edad"))
        if edad < 0:
             print("No es una edad existente.\t Intente de nuevo.")
             #Revisa la condición del while de nuevo
             continue
        else:
            #Edad valida rompe el ciclo y continua con el for.
             break
    #Aplicar descuentos
    descuento = 0
    #condicionantes edades para precio base
    if edad < 3:
        precioBoleto = 0
        print("Boleto gratis")
        continue
        
    elif edad >= 3 and edad <= 17:
        precioBoleto = 30  

        #Puede tener descuento UNICAMENTE DE ESTUDIANTE, por la edad no puede ser profesor ni adulto mayor.
        tipo_visitante = str(input("¿Es estudiante (Ingrese 'si' o 'no'): "))

        #Si es estudiante
        if tipo_visitante == "si":
         descuento = precioBoleto * 0.10
    #Si es mayor de edad edad >= 18:
    else:
        precioBoleto = 45
        tipo_visitante = int(input("Ingrese estudiante:1 profesor:2 o AdultoMayor:3 Ninguno:4): "))

        if tipo_visitante==4:
            print("No hay descuento")

        #Si tiene descuento:
        if tipo_visitante == 3:
            descuento = precioBoleto * 0.12

            print(f"Si es mayor e edad y es Adulto mayor: {precioBoleto}")

        elif tipo_visitante == 1 or tipo_visitante == 2:
            descuento = precioBoleto * 0.10

            print(f"Si es mayor e edad y es {tipo_visitante}: {precioBoleto}")
    #Operacion Si TIENE DESCUENTO
    precioBoleto = precioBoleto - descuento
    print(f"El descuento: {descuento}, el precio del boleto final:{precioBoleto}")
    #Acumular el total de los boletos
    total += precioBoleto
print(total)
```
