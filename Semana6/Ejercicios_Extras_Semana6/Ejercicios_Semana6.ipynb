#### Ejercicio 10: Suma acumulada con `while`
Pide números mientras sean distintos de `0` (el `0` termina el ciclo y no se suma).
**Entrada:** `5`, `8`, `2`, `0` → **Salida:** `La suma de los números es: 15`

````python
suma = 0
while True:
    numero_sum = float(input("El 0 termina el ciclo.\nIngresa número a sumar: "))
    if numero_sum == 0:
        break
    else: 
        suma += numero_sum

print(f"La suma de los números es: {suma}")
````

#### Ejercicio 13: Suma de una lista con función que regresa valor (sin `sum()`)
**Entrada:** `[10, 20, 30, 40]` → **Salida:** `La suma de los elementos es: 100`

````python
def suma_lista(lista):
    suma = 0
    for elemento in lista:
        suma += elemento
    return suma

lista = [10, 20, 30, 40]
print("La suma de los elementos es: ", suma_lista(lista))
   
````

#### Ejercicio 14: Máximo de una lista con función que regresa valor (sin `max()`)
**Entrada:** `[7, 3, 9, 2]` → **Salida:** `El elemento máximo es: 9`

````python
def calcular_numero_max (lista):
    num_max = lista[0]
    for elemento in lista:
        if elemento >= num_max:
            num_max = elemento
    return num_max

lista = [7, 3, 9, 2]

print(f"El elemento máximo es: {calcular_numero_max(lista)}")
````

#### Ejercicio 15: Matriz y suma de la diagonal principal
**Entrada:** `[[1,2,3],[4,5,6],[7,8,9]]` → **Salida:** `La suma de la diagonal principal es: 15`


````python
matriz= [[1,2,3],[4,5,6],[7,8,9]]

suma_diagonal_m = 0

for c in range (len(matriz)):
    for f in range(len(matriz[c])):
        if c == f:
            suma_diagonal_m += matriz[c][f]

print(f"La suma de la diagonal principal es: {suma_diagonal_m}")
````

#### Ejercicio 18: Factorial con función recursiva
**Entrada:** `5` → **Salida:** `El factorial de 5 es: 120`

````python
def factorial_recursivo(num):
    if num== 0 or num== 1:
        return 1
    return num * factorial_recursivo(num - 1)

numero_us=int(input("Ingrese número: "))
print(f"El factorial de {numero_us} es : {factorial_recursivo(numero_us)}")
````

**Entrada:** `(4, 8, 15, 16, 23, 42)` → **Salida:**
```
El tercer elemento es: 15
La suma de la tupla es: 108

#### Ejercicio 21: Diccionario de calificaciones y promedios

```

````python
calificaciones = {"Ana": [8, 9, 10], "Beto": [7, 7, 8]}

for alumno , calif in calificaciones.items():
    suma_calif = 0.00
    promedio = 0.00
    for elemento in calif:
        suma_calif += elemento
    promedio = suma_calif / len(calificaciones[alumno])
  
    print(f"{alumno}: {promedio:.2f}")
````
