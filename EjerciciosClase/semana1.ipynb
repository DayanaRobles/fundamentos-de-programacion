## Ejercicios AUTOevaluables.

```python
tempC= float(input("Ingrese la temperatura en grados Celsius: "))
tempF= (tempC * 9/5) + 32   
tempK= tempC + 273.15
print(f"La temperatura en grados Fahrenheit es: {tempF} y la temperatura en grados Kelvin es: {tempK}")
```

```python
calif1= float(input("Ingrese la primera calificación: "))
calif2= float(input("Ingrese la segunda calificación: "))
calif3= float(input("Ingrese la tercera calificación: "))
promedio= (calif1 + calif2 + calif3) / 3
print(f"El promedio final es: {promedio:.2f}")
```

```python
print("Bienvenido a tu descompositor de tiempo")
seg=int(input("Ingrese los segundos que desea descomponer: "))
tiempo_hr= seg // 3600
tiempo_min= (seg % 3600) // 60
tiempo_seg= seg  % 60
print(f"El tiempo descompuesto es: {tiempo_hr} horas, {tiempo_min} minutos y {tiempo_seg} segundos")
```
