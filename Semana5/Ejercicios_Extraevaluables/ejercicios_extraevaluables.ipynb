### Extra 1: Sistema de calificaciones con tuplas
**Enunciado:** Crea una tupla `calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)`. Imprime la tercera calificación, captura dos calificaciones nuevas con `input()` y anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala **de mayor a menor** y muestra la suma de todos los elementos usando una función que retorne el valor.
**Entrada:**
```
Nueva calificación 1: 8.5
Nueva calificación 2: 9.5
```
**Salida:**
```
Tercera calificación: 8.0
Nueva tupla: (7.5, 9.0, 8.0, 6.5, 10.0, 8.5, 9.5)
Lista ordenada (mayor a menor): [10.0, 9.5, 9.0, 8.5, 8.0, 7.5, 6.5]
Suma total: 59.0
```

### Planteamiento lógico aplicado:

- Usa una tupla para las calificaciones iniciales y la convierte a lista para poder ordenarla.

- Usa ciclos while y try-except para evitar que el programa falle si el usuario ingresa letras en lugar de números.

- La función suma_calif regresa la suma total mediante return.

### Funciones
- suma_calif: recibe una lista, inicia un acumulador en 0, recorre los elementos con un for sumándolos y regresa el total con return.

### Desarrollo del programa
1. Crea la tupla de calificaciones y muestra la tercera usando su índice [2].

2. Pide dos calificaciones nuevas con while y try-except atrapando errores de ValueError.

3. Une la tupla base con las nuevas calificaciones para crear una tupla extendida.

4. Pasa la tupla a lista con list() y la ordena de mayor a menor con .sort(reverse=True).

5. llama a suma_calif pasándole la lista ordenada e imprime el resultado final.

````python
# Suma los elementos de la lista recorriéndola con un ciclo for 
# y regresa el total con return.
def suma_calif(lista):
    suma = 0
    for d in lista:
        suma += d 
    return suma

#Tupla inicial de calificaciones
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

#Captura de nuevas calificaciones con validación directa
#Se usaron ciclos while y try-except para evitar errores de tipo ValueError
print(f"Tercera calificación: {calificaciones[2]}")

#Captura nuevas calificaciones
while True:
        try: 
            calif_e1 = float(input("Nueva calificación 1: "))
            break
        except ValueError:
            print("Error: Ingresé numeros en dígitos únicamente")

while True:
        try:
            calif_e2 = float(input("Nueva calificación 2: "))
            break
        except ValueError:
            print("Error: Ingresé numeros en dígitos únicamente")
            

#Agregar las nuevas calificaciones a una nueva tupla mediante concatenación
calif_exten = calificaciones + (calif_e1, calif_e2)

#Mostrar la nueva tupla extendida
print("Nueva tupla: ", calif_exten)

#Convertir la tupla a lista y ordenarla de mayor a menor con el método .sort()
calif_lis=list(calif_exten)
calif_lis.sort(reverse=True)
print("Lista ordenada (Mayor a menor): ", calif_lis )

#Sumar el total de las calificaciones llamando a la función de suma
resultado = suma_calif(calif_lis)
print("Suma total: ", resultado)





````

**Enunciado:** Crea un diccionario `agenda` con tres contactos iniciales (`"Ana": "555-0101"`, `"Luis": "555-0102"`, `"Mía": "555-0103"`). Captura un nuevo contacto (nombre y teléfono) con `input()` y agrégalo al diccionario. Luego imprime todos los nombres con `keys()` en una sola línea y, con una función que retorne el valor, busca y muestra el teléfono del nuevo contacto.
**Entrada:**
```
Nombre del nuevo contacto: Pedro
Teléfono del nuevo contacto: 555-0140
Nombre a buscar: Pedro
```
**Salida:**
```
Contactos registrados: Ana, Luis, Mía, Pedro
El teléfono de Pedro es: 555-0140
```

### Planteamiento lógico aplicado:

- Usa un diccionario para almacenar los contactos en pares clave-valor (nombre y teléfono).

- Usa un ciclo while con try-except para validar que el número telefónico solo contenga dígitos y guiones.

- Utiliza la función obtener_telefono para consultar el diccionario de forma segura con .get().

### Funciones
- obtener_telefono: recibe el diccionario de contactos y un nombre, le aplica .capitalize() para darle el mismo formato y regresa el teléfono registrado usando .get().

### Desarrollo del programa

1. Inicia el diccionario agenda con tres contactos base.

2. Pide el nombre del nuevo contacto y le da formato con .capitalize().

3. Pide el número telefónico y lo valida en un ciclo while usando .replace("-", "") e int() con try-except para evitar errores.

4. Agrega el nuevo contacto al diccionario agenda.

5. Pide el nombre a buscar, le da formato con .capitalize() y recorre .keys() para imprimir la lista de nombres separados por coma.

6. Usa obtener_telefono para buscar el número del contacto e imprime el resultado.

````python
# Diccionario inicial
agenda = {
    "Ana": "555-0101", 
    "Luis": "555-0102", 
    "Mía": "555-0103"
}

# Función para buscar contacto
# Aplica .capitalize() para mantener un formato
# y con .get() evita excepciones si no existe.
def obtener_telefono(contactos, nombre):
    return contactos.get(nombre.capitalize())

#Pide un nombre nuevo 
nuevo_nombre = input("Nombre del nuevo contacto: ")
# y le da formato con .capitalize()
nombre_extra = nuevo_nombre.capitalize()
# Utiliza un ciclo while con try-except para validar
# que solo se ingresen números y guiones
while True:
    numero = input("Ingrese numero del contacto: ")
    solo_numeros = numero.replace("-", "")
    try:
        numero_int = int(solo_numeros)
        if numero_int >= 0:
            break  
        else:
            print("Error: El número no puede ser negativo.")
    except ValueError:
        print("Error: Ingresa únicamente dígitos numéricos (se permiten guiones).")

# Agrega al diccionario
agenda[nombre_extra] = numero

# Pide el nombre para buscar
nombre_buscar = input("Nombre a buscar: ")
nombre_buscar = nombre_buscar.capitalize()

# Recorre con .keys() 
# e imprime los contactos separados por coma excepto el último
print("Contactos registrados: ", end="")
for n in agenda.keys():
    if n == list(agenda.keys())[-1]:
        print(n) 
    else:
        print(n, end=", ")

        
telefono = obtener_telefono(agenda, nombre_buscar)
print(f"El teléfono de {nombre_buscar} es: {telefono}")
````

### Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con `items()` el color ganador junto con su total.


### Planteamiento lógico aplicado:

- Usa un diccionario para contabilizar y acumular los votos de cada opción.

- Usa un ciclo while y condicionales if-elif-else para validar que solo se cuenten 5 votos válidos.

-  busca con .items() para determinar la opción ganadora o si hubo empate.


### Desarrollo del programa

1. Inicializa el diccionario votos con 0 en cada color y un contador en 1.

2. Registra 5 votos válidos usando while, pasando la entrada a mayúsculas con .upper() para acumular los puntos al color correspondiente.

3. Si ingresa una opción no válida, muestra un mensaje de error y no aumenta el contador.

4. Muestra el resultado total de los votos recorriendo el diccionario con .items().

5. Compara los totales para encontrar el número máximo de votos y detecta si hay colores empatados.

6. Muestra el color ganador o las opciones que empataron.

````python
# Diccionario inicial votos por color
votos = {
    "Rojo": 0,
    "Azul": 0,
    "Verde": 0,
    }

#Contador para numero de votos.
contador = 1

#Ciclo para registrar los 5 votos.
while contador <= 5: 

    voto = input(f"Ingresa el voto {contador}: ")
    # Convertir a mayúsculas para comparar sin errores de formato
    voto_mayusc = voto.upper()
    #Validamos que esté el color que eligió. 
    #si existe, Se suma conforme a su color
    #Si no se repite el turno
    
    if voto_mayusc == "ROJO":
        votos["Rojo"] += 1
        contador += 1
        print("Voto para rojo")
    elif voto_mayusc == "AZUL":
        votos["Azul"] += 1
        contador += 1
        print("Voto para azul")
    elif voto_mayusc == "VERDE":
        votos["Verde"] += 1
        contador += 1
        print("Voto para verde")
    else:
        print("Error: Intenta de nuevo (Rojo, Azul, Verde).")

# Variables para determinar al ganador o si hay empate
max_votos = 0
color_ganador=""
empate=False

#Resultados de los votos
print("\nResultados:")
for k, v in votos.items():
    print(f"Opción {k}, votos: {v}")
    
# Recorrer el diccionario con .items() para buscar al mayor o empates
for color, total in votos.items():
    if total > max_votos:
        max_votos = total
        color_ganador = color
        empate = False
    elif total == max_votos and max_votos > 0:
        color_ganador = color_ganador + " y " + color
        empate = True

# Imprimir al ganador o avisar si hubo empate
if empate:
    print(f"\nHubo un empate entre {color_ganador} con {max_votos} votos cada uno.")
else:
    print(f"\nEl color ganador es {color_ganador} con {max_votos} votos.")
````

### Planteamiento lógico aplicado:

- Cadenas de texto (strings) para modificar texto.

- Usa la función contar_frase que divide la frase y regresa el total de palabras con return.

- Se usa len(), .upper() y .replace() 

### funciones

- contar_frase: recibe un texto, lo separa en una lista de palabras con .split() y regresa la cantidad de elementos usando len()

### Desarrollo del programa

1. Define la función contar_frase y pide al usuario una frase para guardarla en la variable frase.

````python
# Función para contar las palabras de un texto
# Separa la cadena en palabras con .split() 
# y regresa cuántas son con len(

def contar_frase(texto):
    palabras = texto.split()
    return len(palabras)

# Pide la frase al usuario y la guarda
frase = input("Ingresa una frase: ")

# Muestra la longitud total de la frase en caracteres
print("Longitud de la frase: ", len(frase))

# Convierte la frase a mayúsculas con .upper()
frase_mayus = frase.upper()
print("Frase Mayúscula: ", frase_mayus)

# Reemplaza la palabra "Python" por "programación" usando .replace()
frase_modif = frase.replace("Python", "programación")
print("Frase modificada: ", frase_modif)

#Cantidad de palabras
total_palabras = contar_frase(frase)
print("Palabras totales:", total_palabras)
````

````python

# Función auxiliar para sumar elementos de una lista
def suma_calif(lista):
    suma = 0
    for d in lista:
        suma += d 
    return suma

# Función auxiliar para buscar un teléfono en el diccionario
# Usa .capitalize() para darle el formato correcto y .get() para buscar de forma segura
def obtener_telefono(agenda, nombre):
    return agenda.get(nombre.capitalize())

# Función auxiliar para contar las palabras de una frase
def contar_palabras(texto):
    return len(texto.split())

# --- MÓDULOS DE LAS OPCIONES DEL MENÚ ---

# Función Tupla 
def ordenar_tupla():
    print("\n---ORDENAR TUPLA---")
    calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

    #Muestra primero la tercera calificación
    print(f"Tercera calificación: {calificaciones[2]}")
    # Captura nuevas calificaciones con validación try-except
    while True:
        try: 
            calif_e1 = float(input("Nueva calificación 1: "))
            break
        except ValueError:
            print("Error: Ingresé numeros en dígitos únicamente")

    while True:
        try:
            calif_e2 = float(input("Nueva calificación 2: "))
            break
        except ValueError:
            print("Error: Ingresé numeros en dígitos únicamente")

    # Agregar las nuevas calificaciones creando una tupla extendida
    calif_exten = calificaciones + (calif_e1, calif_e2)
    print("Nueva tupla: ", calif_exten)

    #Convertir a lista y ordenar de mayor a menor
    calif_lis=list(calif_exten)
    calif_lis.sort(reverse=True)
    print("Lista ordenada (Mayor a menor): ", calif_lis )

    # Sumar e imprimir total
    resultado = suma_calif(calif_lis)
    print("Suma total: ", resultado)
    
# Función para buscar contacto
def buscar_contacto():
    print("--Agenda--")

    # Diccionario inicial
    agenda = {
        "Ana": "555-0101", 
        "Luis": "555-0102", 
        "Mía": "555-0103"
    }

    #Pide un nombre nuevo 
    nuevo_nombre = input("Nombre del nuevo contacto: ")
    # y le da formato con .capitalize()
    nombre_extra = nuevo_nombre.capitalize()
    # Utiliza un ciclo while con try-except para validar
    # que solo se ingresen números y guiones
    while True:
        numero = input("Ingrese numero del contacto: ")
        solo_numeros = numero.replace("-", "")
        try:
            numero_int = int(solo_numeros)
            if numero_int >= 0:
                break  
            else:
                print("Error: El número no puede ser negativo.")
        except ValueError:
            print("Error: Ingresa únicamente dígitos numéricos (se permiten guiones).")

    # Agrega al diccionario
    agenda[nombre_extra] = numero


    # Recorre con .keys() 
    # e imprime los contactos separados por coma excepto el último
    print("Contactos registrados: ", end="")
    for n in agenda.keys():
        if n == list(agenda.keys())[-1]:
            print(n) 
        else:
            print(n, end=", ")
    
    # Pide el nombre para buscar
    nombre_buscar = input("Nombre a buscar: ")
    nombre_buscar = nombre_buscar.capitalize()

    telefono = obtener_telefono(agenda, nombre_buscar)
    print(f"El teléfono de {nombre_buscar} es: {telefono}")

# Funcion Division con excepciones
def dividir_numeros():
    print("\n---DIVIDIR NUMEROS---")
    while True:
        try:
            num1 = int(input("Ingrese el primer número entero: "))
            break
        except ValueError:
            print("Error: Ingresa un número entero válido")
    
    while True:
        try:
            num2 = int(input("Ingrese el segundo número entero: "))
            break
        except ValueError:
            print("Error: Ingresa un número entero válido")
    
    try:
        resultado = num1 / num2
        print(f"Resultado: {num1} / {num2} = {resultado}")
    except ZeroDivisionError:
        print("Error: No se puede dividir entre cero.")

# Funcion analizar texto
def analizar_texto():
    
    print("--Texto--")
    
    #Mensaje original
    mensaje = "Python es un lenguaje poderoso"

    #Mostrar longitud mensaje usando len()
    print("Longitud del mensaje:", len(mensaje))

    #Convertir a mayúsculas usando .upper()
    mensaje_mayus= mensaje.upper()
    print(f"En mayúsculas: {mensaje_mayus}")

    #Reemplazar una palabra usando .replace()
    mensaje_modificado = mensaje.replace("Python", "programación")
    print(f"Texto reemplazado: {mensaje_modificado}")

    #Contar palabaras apartir de la función
    total_palabras = contar_palabras(mensaje)
    print("Palabras totales:", total_palabras)


# Menu Integrador
opcion = 0
while opcion != 5:
    print("\n--MINIMENÚ MODULAR INTEGRADOR---")
    print("1: Mostrar tupla de calificaciones ordenada")
    print("2: Buscar un teléfono en contactos")
    print("3: Dividir dos números")
    print("4: Analizar un mensaje de texto")
    print("5: Salir")

    try:
        opcion =int(input("Elige una opción (1-5): "))
    except ValueError:
        print("Error: Ingresa el número en dígito (1, 2, 3, 4 o 5).")
        continue

    if opcion == 1:
        ordenar_tupla()
    elif opcion == 2:
        buscar_contacto()
    elif opcion == 3:
        dividir_numeros()
    elif opcion == 4:
        analizar_texto()
    elif opcion == 5:
        print("\nCerrando el programa...")
    else:
        print("Opción no válida. Elige un número entre 1 y 5.")



        
````
