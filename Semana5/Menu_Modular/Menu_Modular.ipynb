## Planteamiento lógico aplicado:

- En el programa se utilizaron tuplas, listas, diccionarios y cadenas de texto (strings).

- Mediante ciclos while y bloques try-except se evita que el programa falle por datos inválidos (ValueError, ZeroDivisionError)

- Se dividió el programa en funciones auxiliares reutilizables y funciones para cada opción del menú.

### Funciones auxiliares.

1. suma_calif: recibe una lista de calificaciones y empieza una suma desde 0. Después recorre la lista con un for y va sumando cada calificación. Al final regresa el resultado con return.

2. pedir_calif: pide al usuario una calificación y usa while y try-except para revisar que el dato sea correcto. También verifica que la calificación esté entre 0 y 100. Si hay un error, vuelve a pedirla.

3. obtener_telefono: recibe un diccionario y un nombre. Usa .get() para buscar el nombre y .capitalize() para darle el formato correcto. Si encuentra el contacto, regresa su teléfono; si no, regresa None.

4. pedir_entero: pide un número entero al usuario. Usa while y try-except para evitar errores cuando se escribe algo que no sea un entero.

5. contar_palabras: recibe un texto y lo separa en palabras usando .split(). Después utiliza len() para saber cuántas palabras hay.

### Módulos del menú
- opcion_tupla: crea una tupla con algunas calificaciones y después pide dos más usando pedir_calif. Junta todas las calificaciones en otra tupla, muestra una usando su índice. Se convierte a lista para poder ordenarla de mayor a menor y al final, calcula la suma con suma_calif.

- opcion_diccionario: crea un diccionario con algunos contactos. Después pide un nuevo nombre y teléfono, revisando que el teléfono tenga números. Agrega el contacto al diccionario, muestra los nombres guardados y busca un teléfono usando obtener_telefono.

- opcion_excepciones: pide dos números enteros usando pedir_entero. Primero muestra la suma y después intenta dividirlos. Si el segundo número es 0, utiliza try-except para evitar el error de división entre cero.

- opcion_strings: crea un texto y realiza diferentes operaciones con él. Obtiene su longitud con len(), lo convierte a mayúsculas con .upper(), cambia una palabra con .replace() y cuenta las palabras usando contar_palabras.

### Menú principal
- menu_principal: muestra las 5 opciones del programa dentro de un ciclo while True. Con try-except revisa que la opción ingresada sea correcta y con if-elif-else decide qué función ejecutar. Cuando el usuario elige la opción 5, utiliza break para salir del ciclo.

- Punto de partida: al final del programa se llama a menu_principal() para comenzar a ejecutar el menú.

```python
#Funciones auxiliares

#Función auxliar sumar calificaciones:
def suma_calif(lista):
    suma = 0
    for d in lista:
        suma += d 
    return suma

#Función auxliar pedir calificaciones:
def pedir_calif():
    #Pide una calificación y no deja avanzar hasta que sea un número válido entre 0 y 100.
    # El try-except para que si el usuario mete letras, el programa no crashee
    while True:
        try:
            num = float(input("Ingrese calificación: "))
            if 0 <= num <= 100:
                return num
            else:
                print("Error: La calificación debe estar entre 0 y 100.")
            
        except ValueError:
            print("Error: Ingresa un número válido")

#Función auxliar obtener teléfono:
#Busca un teléfono en el diccionario. 
#Usa .capitalize() para poder hacer la comparación sin errores de formato
#y .get() para que si no existe el contacto devuelva None en vez de lanzar un error.
def obtener_telefono(contactos, nombre):
    return contactos.get(nombre.capitalize())

# Función pedir numero entero 
# Validar con try-except que no entren letras ni decimales
def pedir_entero():
    while True:
        try:
            num = int(input("Ingrese un número entero: "))
            return num
        except ValueError:
            print("Error: Ingresa un número entero válido")

#Función para contar palabras:
#Cortamos el texto en palabras con .split() 
# y cuenta cuántas son usando len()
def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)
#---------------Funciones separadas Menú------------------
# Función menú tupla: ordenar
def opcion_tupla():

    #ENCABEZADO
    print("\n---TUPLA CALIFICACIONES---")
    #Calificaciones iniciales
    calificaciones_grupo = (85, 92, 78, 95, 88, 70, 100)
    print(f"Calificaciones originales: {calificaciones_grupo}")

    #Pedir dos califiaciones: usando la función auxiliar
    # Y las guarda
    calif_e1= pedir_calif()
    calif_e2= pedir_calif()
            
    # Agrega las nuevas calificaciones a una nueva tupla
    calif_exten = calificaciones_grupo + (calif_e1, calif_e2)

    # Muestra la 3ra calificación que en la tupla tiene subíndice 2
    print(f"Tercera calificación: {calificaciones_grupo[2]}")

    # Mostrar nueva tupla con las calificaciones añadidas:
    print("Nueva tupla: ", calif_exten)

    #Convertimos a lista para ordenar de mayor a menor con .sort()
    calif_lis=list(calif_exten)
    calif_lis.sort(reverse=True)
    print("Lista ordenada (Mayor a menor): ", calif_lis )

    # Sumamos el total de las calificaciones 
    # usando la función auxiliar de suma
    resultado = suma_calif(calif_lis)
    print("Suma total: ", resultado)

# Función menú diccionario
def opcion_diccionario():

    #ENCABEZADO
    print("\n---DICCIONARIO---")

    #Directorio inicial   
    directorio_alumnos = {
        "Ana": "555-1234",
        "Carlos": "555-5678",
        "Sofia": "555-8765"
    }

    # Agregar nuevo contacto
    #Pedir nombre 
    nuevo_nombre = input("Ingresa el nombre del nuevo contacto: ")

    #Guardar en formato con mayúscula incial
    nombre_nuevo_formato= nuevo_nombre.capitalize()

    # Validar número de teléfono: solo dígitos númericos
    #usamos el .replace() para cambiar los guiones en caso
    # de que el formato sea 121-1293
    while True:
        nuevo_telefono = input("\nIngrese número del contacto: ")
        solo_numeros = nuevo_telefono.replace("-", "")
        #No aceptar números negativos.
        try:
            numero_int = int(solo_numeros)
            if numero_int >= 0:
                break  
            else:
                print("Error: El número no puede ser negativo.")
       
        except ValueError:
            print("Error: Ingresa únicamente dígitos numéricos (se permiten guiones).")

    # Agregar el telefono ya válido
    directorio_alumnos[nombre_nuevo_formato] = nuevo_telefono

    #Imprimir los nombres recorriendo las .keys()
    print("\nNombres de alumnos registrados: ")
    for nombre_contacto in directorio_alumnos.keys():
        print(f"- {nombre_contacto}")

    #Pedir el nombre para buscar
    nombre_busqueda = input("Ingresa el nombre del alumno a buscar: ")
    nombre_busqueda_formato = nombre_busqueda.capitalize()

    #Buscar el teléfono con la función auxiliar .get()
    telefono = obtener_telefono(directorio_alumnos, nombre_busqueda_formato)
    #Mostrar si el teléfono está registrado
    if telefono: 
        print(f"El teléfono de {nombre_busqueda_formato} es: {telefono}")
    else:
        print("El alumno no se encuentra registrado en el directorio.")

# Funcion menú: Excepciones

def opcion_excepciones():
    #ENCABEZADO
    print("\n---EXCEPCIONES---")

    #Pedir numeros para división y validar con función aux de pedir enteros
    # La función auxiliar valida que sea enteros.
    print("Primer número (dividendo):")
    num1 = pedir_entero()

    print("Segundo número (divisor):")
    num2 = pedir_entero()

    #Sumar ambos numeros
    suma = num1 + num2
    print(f"Suma: {num1} + {num2} = {suma}")

    
    # Intentar la división, mostrar error si el divisor es 0
    try:
        resultado = num1 / num2
        print(f"Resultado: {num1} / {num2} = {resultado}")
    except ZeroDivisionError:
        # Si el divisor es 0 mostrar mensaje de error.
        print("Error: No se puede dividir entre cero.")

# Funcion menú: Strings
def opcion_strings():
    #ENCABEZADO
    print("\n---Strings---")


    mensaje = "Me gusta estudiar diseño"
    print(f"Mensaje original: {mensaje}")

    # longitud del mensaje usando len()
    print(f"Longitud del mensaje: {len(mensaje)}")

    # Convertir a Mayus usando .upper()
    mensaje_mayus=mensaje.upper()
    print(f"En mayúsculas: {mensaje_mayus}")

    #Reemplazar una palabra usando .replace()
    mensaje_modificado = mensaje.replace("diseño", "programación" )
    #Mostrar el nuevo mensaje
    print(f"Texto reemplazado: {mensaje_modificado}")

    # Contar palabras usando la función auxiliar 
    total =contar_palabras(mensaje)
    print(f"Cantidad de palabras: {total}")

#-----------------------Menu principal-------------------------

def menu_principal():
    #En ejecución hasta que elija la opción 5
    while True:
        #ENCABEZADO
        print("\n--MINI MENÚ MODULAR INTEGRADOR---")

        #OPCIONES
        print("1: TUPLA")
        print("2: DICCIONARIO")
        print("3: EXCEPCIONES")
        print("4: STRINGS")
        print("5: FINALIZAR")

        #Validar que la opción sea un entero 
        try:
            opcion =int(input("Elige una opción (1-5): "))
        except ValueError:
            print("Error: Ingresa el número en dígito (1, 2, 3, 4 o 5).")
            continue

        if opcion == 1:
            opcion_tupla()
        elif opcion == 2:
            opcion_diccionario()
        elif opcion == 3:
            opcion_excepciones()
        elif opcion == 4:
            opcion_strings()
        elif opcion == 5:
            print("\nCerrando el programa...")
            print("\nAdios!")
            break
        else:
            print("Opción no válida. Elige un número entre 1 y 5.")

#Punto de partida
menu_principal()
```
