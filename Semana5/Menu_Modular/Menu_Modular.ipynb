Crear una **aplicación interactiva en Python** que demuestre el uso integral de **tuplas**, **diccionarios**, **manipulación de cadenas de texto (strings)** y **manejo robusto de excepciones**, estructurada a través de un **menú principal modular** controlado por un ciclo. El usuario debe poder seleccionar por número qué sección desea ejecutar (Tuplas, Diccionarios, Excepciones, Strings o Finalizar) y el programa debe delegar cada tarea a funciones modulares.

Este reto integra la creación y manipulación de **tuplas** (Tema 13), los **diccionarios** con claves y valores (Tema 14), los bloques **try-except** con excepciones específicas (Tema 15) y los **métodos de strings** con conteo de palabras (Tema 16), además de consolidar las funciones con y sin retorno vistas en semanas anteriores.


```python
calificaciones_grupo = (85, 92, 78, 95, 88, 70, 100)

directorio_alumnos = {
    "Ana": "555-1234",
    "Carlos": "555-5678",
    "Sofia": "555-8765"
}

#---------------Funciones separadas Menú------------------
# Función menú tupla: ordenar
def opcion_tupla():
    print("---TUPLA---")
    print("Calificaciones originales: ", calificaciones_grupo)
    calificaciones_grupo_list = list(calificaciones_grupo)
    calificaciones_grupo_list.sort(reverse=True)
    print("Lista ordenada (Mayor a menor): ", calificaciones_grupo_list)

# Función menú diccionario
def opcion_diccionario():
    print("\n---DICCIONARIO---")
    # Pedir el nombre para buscar
    nombre = input("Ingresa el nombre del alumno a buscar: ")

    # Variable booleana para saber si encontramos al alumno
    encontrado = False
    # Ciclo para buscar itilizando items()
    for alumno, telefono in directorio_alumnos.items():
        # Upper() parac omparar sin importar mayús o minus
        if alumno.upper() == nombre.upper():
            print(f"El teléfono de {alumno} es: {telefono}")
            encontrado = True
            break
    #Si no lo encuentra entonces se muestra este mensaje
    if not encontrado:
        print("El alumno no se encuentra registrado en el directorio.")

# Función pedir numero entero 
def pedir_entero():
    while True:
        try:
            num = int(input("Ingrese un número entero: "))
            return num
        except ValueError:
            print("Error: Ingresa un número entero válido")

# Funcion menú: Excepciones
def opcion_excepciones():
    print("\n---EXCEPCIONES---")
    print("Primer número (dividendo):")
    num1 = pedir_entero()
    print("Segundo número (divisor):")
    num2 = pedir_entero()

    # División
    try:
        resultado = num1 / num2
        print(f"Resultado: {num1} / {num2} = {resultado}")
    except ZeroDivisionError:
        # Si el divisor es 0 mostrar mensaje de error.
        print("Error: No se puede dividir entre cero.")

def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)

def opcion_strings():
    print("\n---Strings---")
    mensaje = "Me gusta estudiar diseño"
    print(f"Mensaje original: {mensaje}")
    # longitud del mensaje
    print(f"Longitud del mensaje: {len(mensaje)}")
    # Convertir a Mayus
    mensaje_mayus=mensaje.upper()
    print(f"En mayúsculas: {mensaje_mayus}")
    #Reemplazar una palabra:
    mensaje_modificado = mensaje.replace("diseño", "programación" )
    print(f"texto reemplazado:{mensaje_modificado}")
    # Contar palabras llamando a la función con retorno
    total =contar_palabras(mensaje)
    print(f"Cantidad de palabras: {total}")

def menu_principal():
    while True:
        print("\n--MINIMENÚ MODULAR INTEGRADOR---")
        print("1: Mostrar tupla de calificaciones ordenada")
        print("2: Buscar un teléfono en contactos")
        print("3: Dividir dos números")
        print("4: Analizar un mensaje de texto")
        print("5: Salir")

        try:
            opcion =int(input("Elige una opción (1-5): "))
        except ValueError:
            print("Error: Ingresa el número en dígito (1, 2, 3, 4 o 5).")
            continue

        if opcion == 1:
            opcion_tupla()
        elif opcion == 2:
            opcion_diccionario()
        elif opcion == 3:
            opcion_excepciones()
        elif opcion == 4:
            opcion_strings()
        elif opcion == 5:
            print("\nCerrando el programa...")
            break
        else:
            print("Opción no válida. Elige un número entre 1 y 5.")

menu_principal()
```
