## GUIA PROYECTO ADELANTO

```python
#Costos de los servicios

c_cambiosuela = 100
c_costuras = 300
c_tintado = 500
c_res_gam = 900
c_ajuste = 20 

#Tiempo de entrega en días por un par.

t_cambiosuela = 3
t_costuras = 5
t_tintado = 7
t_res_gam = 9
t_ajuste = 1

# Variables que cambian con el registro

subtotal = 0
t_entregafinal = 0
m_descuento = 0
m_express = 0 

# Contadores individuales para cada servicio y afinar tiempos de entrega.
cont_suela = 0
cont_costura = 0
cont_tintado = 0
cont_gamuza = 0
cont_ajuste = 0

#INICIO PROGRAMA

cliente=input("Ingrese nombre completo del cliente: ")

#Verificar que sea válida la cantidad de servicios por ticket

while True:
    c_pedidos = int(input("Ingrese cantidad de servicios a requerir por el cliente (Máximo 10): "))
    
    if 1 <= c_pedidos and c_pedidos <= 10:
        break
    else:
        print("La cantidad de servicios debe ser entre 1 y 10 por ticket.\nPor favor, intente nuevamente.")

#Despligue menu de servicios

print("\n=======MENU========")
print("========DE=========")
print("=====SERVICIOS=====\n")
print(" (1)Cambio Suela\n (2)Costuras\n (3)Tintado\n (4)Restauración gamuza\n (5)Ajuste de correa")

#Inicia resgistro de pedidos.
print("\n=======================")
print("=========Ticket========")
print("=======================")
print(f"Nombre cliente: {cliente}")


for pedido in range (1, c_pedidos + 1):

    #Verificar que exista la opición seleccionada: 
    #si existe entonces verificar si ya llegó al limite por servicios solicitados. 
    
    while True:
        
        tiposervicio=int(input(f"Ingrese número de servicio no.{pedido}: "))
        
        if tiposervicio >= 1 and tiposervicio <= 5:
          
            if tiposervicio == 1 and cont_suela >= 3:
                print("Se llegó al límite de cantidad de servicios posibles en -Cambio de Suela-.\nElija otro servicio.")
        
            elif tiposervicio == 2 and cont_costura >= 3:
                print("Se llegó al límite de cantidad de servicios posibles en -Costuras-.\nElija otro servicio.")
        
            elif tiposervicio == 3 and cont_tintado >= 2:
                print("Se llegó al límite de cantidad de servicios posibles en -Tintado-.\nElija otro servicio.")
        
            elif tiposervicio == 4 and cont_gamuza >= 2:
                print("Se llegó al límite de cantidad de servicios posibles en -Restauración de Gamuza-.\nElija otro servicio.")
        
            elif tiposervicio == 5 and cont_ajuste >= 9:
                print("Se llegó al límite de cantidad de servicios posibles en -Ajuste de correas-.\nElija otro servicio.")
                
            else: 
                break
        else:
            print("Opción no válida. Seleccione un servicio del 1 al 5.")              

    #Procesos internos dependiendo del servicio: 

    if tiposervicio == 1:
        print(f"Costo servicio 'Cambio de suela': ${c_cambiosuela:.2f}")
        subtotal += c_cambiosuela
        cont_suela += 1

    #Regla: Tiempo base por un par es 3 días, por cada par extra 2 días más

        t_servicio = t_cambiosuela + (cont_suela - 1) * 2
        if t_servicio > t_entregafinal:
            t_entregafinal = t_servicio

    elif tiposervicio == 2:
        print(f"Costo servicio 'Costura': ${c_costuras:.2f}")
        subtotal += c_costuras
        cont_costura += 1 

    #Regla: Tiempo base por un par es 5 días, por cada par extra 3 días más
        
        t_servicio = t_costuras + (cont_costura - 1) * 3
        if t_servicio > t_entregafinal:
            t_entregafinal = t_servicio    

    elif tiposervicio == 3:
        print(f"Costo servicio 'Tintado': ${c_tintado:.2f}")
        subtotal += c_tintado
        cont_tintado  += 1

     #Regla: Tiempo base por un par es 7 días, por cada par extra 4 días más
        
        t_servicio = t_tintado + (cont_tintado - 1) * 4
        if t_servicio > t_entregafinal:
            t_entregafinal = t_servicio

    elif tiposervicio == 4:
        print(f"Costo servicio 'Restauración Gamuza': ${c_res_gam:.2f}")
        subtotal += c_res_gam
        cont_gamuza += 1

    #Regla: Tiempo base por un par es 9 días, por cada par extra 5 días más
                
        t_servicio = t_res_gam + (cont_gamuza - 1) * 5
        if t_servicio > t_entregafinal:
            t_entregafinal = t_servicio

    
    elif tiposervicio == 5:
        print(f"Costo servicio 'Ajuste Correa': ${c_ajuste:.2f}")
        subtotal += c_ajuste
        cont_ajuste += 1
       
        #Mismo día
        if cont_ajuste <= 2:
            t_servicio= 0  
        #Día siguiente
        elif cont_ajuste <= 5:
            t_servicio = t_ajuste 
        # A partir de 6 pares: Tiempo base 1 día, por cada par extra 1 día más.
        else:
            t_servicio = t_ajuste + (cont_ajuste - 5) * 1
            
        if t_servicio > t_entregafinal:
            t_entregafinal = t_servicio
    
    
#Termina registro servicios.
print(f"\nSubtotal: ${subtotal:.2f}")
#Muestra si aplica descuento (10%) y si no el monto de descuento es 0

if c_pedidos > 2:
    print("Descuento aplicado: 10%")
    m_descuento = subtotal * 0.10
else:
    m_descuento=0    

#Pregunta si el pedido es urgente
#solo aplica cuando el tiempo de entrega es mayor a un día

if t_entregafinal > 1: 
    while True:
        print("\n¿servicio Express?")
        urgente = input("Ingrese 'si' o 'no': ")
        #Verificar que sea válida la opcion
        if urgente == "si" or urgente == "Si" or urgente == "SI" or urgente == "no" or urgente == "No" or urgente == "NO":
            break
        print("Respuesta no válida. Ingrese 'si' o 'no'")

    #si es urgente aplicar el 20% de recargo y reducir el tiempo de entrega
    #Únicamente días enteros.
    
    if urgente == "si" or urgente == "Si" or urgente == "SI":
        m_express = subtotal * 0.20  
        t_entregafinal = t_entregafinal // 2
        print("Servicio Express aplicado: +20% ")
    else:
        print("Servicio Express no solicitado.")
        m_express=0
#Si la entrega es para el mismo día, en caso de que sea ajuste de correa si es unitaria puede ser el mismo día cuando se entregue.
else:
    urgente= "no"
    m_express = 0
    print("El pedido no aplica para Servicio Express.")
       
#Operación total: se aplican descuentos y recargos.
total = subtotal - m_descuento + m_express

#Mostrar sí hay descuento o Recargos, y sus respectivos montos

if m_descuento > 0:
    print(f"Monto descuento: ${m_descuento:.2f}")
    
if m_express > 0:
    print(f"Monto Servicio Express: ${m_express:.2f}")

#Mostrar total

print(f"\nTOTAL A PAGAR: ${total:.2f}")

#Tiempos de entrega

if t_entregafinal == 0:
    print("Tiempo de entrega: mismo día.")
else: 
    print("Tiempo estimado de entrega: ",t_entregafinal, "días después de anticipo")

#Anticipos: Puede ser ingresado por el cliente
#Reglas: Minimo 50% y Máximo el TOTAL del pedido

#Operación anticipo
anti_min = total * 0.50

#El cliente puede decidir el anticipo: debe ser mínimo el 50% o dejar liquidado.
while True:
    anticipo = float(input(f"Ingrese el anticipo que desea dejar el cliente (Mínimo 50%, equivale a ${anti_min}.\nAnticipo cliente: $"))

    if anticipo >= anti_min and anticipo <= total:
        break
    elif anticipo > total:
        print(f"El anticipo no puede ser mayor al total: ${total:.2f}")
    else: 
        print(f"El anticipo debe ser al menos del 50%: ${anti_min:.2f}")

#Calcular el saldo pediente
s_pendiente= total-anticipo

#Mostrar cantidad del anticipo y restante a pagar.
print(f"Anticipo pagado: ${anticipo:.2f}")

if anticipo == total:
    print("Pedido liquidado.")
else:
    print(f"Restante a pagar: ${s_pendiente:.2f}")

print("\n========¡Buen día!========\nGracias por elegir 'RENOVACALZADO'")
    
```

```python
import pdb

servicios = {

    "Cambio De Suela" : 100,
    "Costura" : 300,
    "Tintando" : 500,
    "Restauración" : 900,
    "Ajuste" : 20,
}

dic_tiempo_entrega = { #[t_base, t_extra, límite]
    "Cambio De Suela" : [3, 2, 3 ],
    "Costura" : [5, 3 , 3],
    "Tintando" : [7, 4, 2],
    "Restauración" : [9, 5, 2],
    "Ajuste" : [1, 1, 9]
}

def mostrar_menu(diccionario_servicios):
    print("\n=======MENU========")
    print("========DE=========")
    print("=====SERVICIOS=====\n")
    servicios_lista = list(diccionario_servicios.keys())
    for i, (nombre, costo) in enumerate(diccionario_servicios.items(), start=1):
        print(f" ({i}) {nombre} - ${costo:.2f}")
    return servicios_lista

def pedir_entero(mensaje, minimo, maximo):
    while True:
        try:
            eleccion_cliente = int(input(f"{mensaje}.Rango{minimo}-{maximo}"))
            
            if eleccion_cliente >= minimo and eleccion_cliente <= maximo:
                return eleccion_cliente
            else:
                print(f"El valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Ingresa un número entero válido: solo valor numérico")

def calcular_tiempo_entrega(servicio, t_entrega, cantidad):
    t_base = dic_tiempo_entrega[servicio][0]
    t_extra= dic_tiempo_entrega[servicio][1]

    if servicio == "Ajuste":
         # Mismo día
        if cantidad <= 2:
            t_servicio= 0  
        #Día siguiente
        elif cantidad <= 5:
            t_servicio = t_base
        else:
            t_servicio = t_base + (cantidad - 5) * t_extra
    else:         
        t_servicio = t_base + (cantidad - 1) * t_extra

    # Comparar tiempo actual acumulado del ticket
    if t_servicio > t_entrega:
        t_entrega = t_servicio
        return t_servicio
    else:
        return t_entrega
   
def opcion_cotización():
    print("# TICKET RENOVACALZADO #")
    cliente = input("Ingrese nombre completo del cliente: ")
    c_pedidos = pedir_entero("Ingrese cantidad de servicios.", 1, 10)
    servicios_opciones = mostrar_menu(servicios)

    print("\n=======================")
    print("=========Ticket========")
    print("=======================")
    print(f"Nombre cliente: {cliente}")

    ticket_servicios=[]
    subtotal=0
    t_entrega = 0
    m_descuento = 0
    #Iteración de pedidos.
    for pedido in range(1, c_pedidos +1):
        #Ingresar servicio y verifica con pedir entero. Límite 1-5
        while True:
            servicio_elegido = pedir_entero("Ingrese Servicio.", 1, 5)
            servicio_real = servicios_opciones[servicio_elegido-1]
            limite = dic_tiempo_entrega[f"{servicio_real}"][2]

            if ticket_servicios.count(servicio_real) < limite:
                print("No supera el limite")
                break
            else:
                print("Supero el límite")

        #Sumar
        ticket_servicios.append(servicio_real)
        subtotal += servicios.get(servicio_real)
        
        # Calcular tiempos de entrega actualizando t_entrega
        cantidad_servicio = ticket_servicios.count(servicio_real)
        t_entrega = calcular_tiempo_entrega(servicio_real, t_entrega, cantidad_servicio)

    print(f"\nSubtotal acumulado: ${subtotal:.2f}")
    ##-----Termina captura de servicios de manera independiente----
    ##-----Inician cálculos para ticket: descuento, tiempo de entrga final, servicio express----
    #Descuentos.
    if c_pedidos > 2:
       print("Descuento aplicado: 10%")
       m_descuento = subtotal * 0.10
    else:
       m_descuento = 0
    # Servicio Express
    m_express = 0
    if t_entrega > 1:
        while True:
            urgente = input("\n¿Servicio Express? (Ingrese 'si' o 'no'): ").strip().lower()
            if urgente == "si" or urgente == "no":
                break
            print("Respuesta no válida. Ingrese 'si' o 'no'.")

        if urgente == "si":
            m_express = subtotal * 0.20
            t_entrega = t_entrega // 2
            print("Servicio Express aplicado: +20%")
        else:
            print("Servicio Express no solicitado.")
    else:
        print("El pedido no aplica para Servicio Express (entrega rápida o mismo día).")

    # Operación total
    total = subtotal - m_descuento + m_express

    if m_descuento > 0:
        print(f"Monto descuento: ${m_descuento:.2f}")
    if m_express > 0:
        print(f"Monto Servicio Express: ${m_express:.2f}")

    print(f"\nTOTAL A PAGAR: ${total:.2f}")

    # Tiempos de entrega finales
    if t_entrega == 0:
         print("Tiempo de entrega: mismo día.")
    else:
        print(f"Tiempo estimado de entrega: {t_entrega} días después de anticipo.")

    # Anticipo mínimo 50%, SÍ puede dejar liquidado
    anti_min = total * 0.50
    while True:
        try:
            anticipo = float(input(f"Ingrese el anticipo (Mínimo 50%, equivale a ${anti_min:.2f}): $"))
            if anti_min <= anticipo <= total:
                break
            elif anticipo > total:
                print(f"El anticipo no puede ser mayor al total: ${total:.2f}")
            else:
                print(f"El anticipo debe ser al menos del 50%: ${anti_min:.2f}")
        except ValueError:
            print("Ingrese un valor numérico válido para el anticipo.")

    s_pendiente = total - anticipo
    print(f"Anticipo pagado: ${anticipo:.2f}")

    if anticipo == total:
        print("Pedido liquidado.")
    else:
        print(f"Restante a pagar: ${s_pendiente:.2f}")

    print("\n========¡Buen día!========\nGracias por elegir 'RENOVACALZADO'")

def main(): 
    while True:
        print("\n=====================")
        print("\n===Menú Principal====")
        print("\n=====================")

        matriz_menu_principal = [
                ["(1) Cotizar Servicio", "(2) Leer Archivos"],
                ["(3) Escribir/anexar documentos", "(4) Salir"]
            ]
        
        for fila in matriz_menu_principal:
            for columna in fila:
                print(columna)
        
        
        opcion_menu = pedir_entero("¿Qué desea hacer? Eliga una opción: ", 1, 4)
        pdb.set_trace()
        # Opción cotizar
        if opcion_menu == 1:
            opcion_cotización()
        # Opción leer archivos
        elif opcion_menu == 2:
            print("\n[Opción 2 seleccionada: Leer archivo de texto]")
        # Opción escirbir archivos
        elif opcion_menu == 3:
            print("\n[Opción 3 seleccionada: Escribir o anexar datos]")
        # Cerrar 
        elif opcion_menu == 4:
            print("Cerrando sesión. ¡Hasta luego!")
            break
main()
```
