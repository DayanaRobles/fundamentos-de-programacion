## Extra 1: División de cuenta con propina
**Enunciado:** Crea un programa que pida el total de la cuenta de un restaurante,
el porcentaje de propina a dejar y el número de personas que pagarán. El programa
debe calcular el monto de la propina, el total a pagar con propina y cuánto le toca
pagar a cada persona (con dos decimales).


### Algoritmo Pseudocódigo.

Escribir "Ingrese la cantidad a pagar"

Leer subtotal

Escribir "ingrese el porcentaje de propina que desea dejar"

Leer porcentaje

Escribir "Ingrese la cantidad de personas que pagaran"

Leer numero_de_personas


propina=subtotal * porcentaje / 100 

total= subtotal + propina

total_por_persona= total / numero_de_personas 

// Resultado redondeado a dos decimales


Escribir "propina:", propina 


Escribir "total con propina", total 


Escribir "pago por persona:", total_por_persona 


finAlgoritmo


```python
subtotal=float(input("Ingrese la cantidad a pagar"))
porcentaje=float(input("ingrese el porcentaje de propina que desea dejar"))
numero_de_personas=int(input("Ingrese el numero de personas que van a pagar la cuenta"))
propina=subtotal*porcentaje/100
total=subtotal+propina
total_por_persona=total/numero_de_personas
print(f"Propina: {propina:.2f}")
print(f"Total con propina: {total:.2f}")
print(f"pago por persona: {total_por_persona:.2f}")
```

## Extra 2. Conversor de minutos a días, horas y minutos.
**Enunciado:** Crea un programa que pida una cantidad total de minutos (entero) y la convierta a días, horas y minutos restantes. Utiliza los operadores de división entera `//` y módulo `%`. (Pista: 1 día = 1440 minutos, 1 hora = 60 minutos.).


### Algoritmo Pseudocódigo.

Escribir "Ingrese la cantidad total de minutos"

Leer minutos

dias=minutos // 1440

horas= (minutos % 1440)// 60

min_rest= minutos % 60

Escribir minutos, "minutos equivalen a: ", dias,"día(s), ", horas, "hora(s), ", min_rest, "minutos(s)"

finAlgoritmo


```python
minutos=int(input("Ingrese la cantidad total de minutos: "))
dias=minutos//1440
horas=(minutos%1440)//60
min_rest=minutos%60
print(f"{minutos} minutos equivalen a {dias} dia(s), {horas} hora(s) y {min_rest} minuto(s)")
```
