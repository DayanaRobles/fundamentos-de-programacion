## La "Tabla de Verdad" de la Actividad 2
Para que **solo se aplique un descuento por boleto**, se evalúan los tipos en orden de prioridad y la primera condición que se cumpla "gana" el descuento:

| ¿Mayor de edad? | Tipo de visitante | Descuento | Boleto |
| :---: | :--- | :---: | :---: |
| Sí | Adulto mayor (12%) | 12% | `45 - 45*0.12 = 39.60` |
| Sí | Profesor (10%) | 10% | `45 - 45*0.10 = 40.50` |
| Sí | Estudiante (10%) | 10% | `45 - 45*0.10 = 40.50` |
| Sí | Otro | 0% | `45.00` |
| No | Menor de 3 años | Boleto gratis | `0.00` |
| No | De 3 a 17 años | 0% | `30.00` |


````python
#Ejercicio

precio=45
edad=45
tipo="estudiante"
if edad < 3:
    print("tu boleto es gratis")
else:
    if tipo== "adulto_mayor":
        descuento = precio * 0.12          
    elif tipo== "profesor" or tipo== "estudiante":
        descuento = precio * 0.10         
    else:
        descuento = 0

    total = precio - descuento
print(f"{tipo}, tu precio final es: ${total:.2f}")
````

### Ejercicio 17: Suma de pares con continue
**Enunciado:** Crea un programa que capture 5 números y sume únicamente los pares, ignorando los impares con `continue`.
**Entrada:** `1`, `2`, `3`, `4`, `5`
**Salida:** `La suma de los pares es: 6`

````python
i=1
suma_numero=0
while i <= 5:
    num=int(input("Ingrese numero {i}"))
    if (num % 2 != 0):
        print("Es numero impar")
        i+=1
        continue
    else:
        suma_numero+=num
        print("numero par")
        i+=1
print(f"La suma de los pares es: {suma_numero}")
````

### Ejercicio 18: Detener la suma con break
**Enunciado:** Crea un programa que sume números capturados con `input()`; si el acumulado alcanza o supera 100, el ciclo se detiene con `break` y se muestra el total.
**Entrada:** `50`, `30`, `25`
**Salida:** `Total: 105`

````python
num_acum=0
while True:
    num_cap=int(input("Ingrese numero: "))
    num_acum += num_cap
    if num_acum >=100:
        break
print("total: ",num_acum)
````
