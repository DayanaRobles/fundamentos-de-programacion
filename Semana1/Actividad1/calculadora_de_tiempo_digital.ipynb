```python
#Pedir el nombre de usuario con la función input() y guardarlo en la variable nombreusuario.
nombreusuario=input("Ingrese su nombre de usuario: ")
#Pedir el tiempo que ha escuchado en diferentes plataformas en horas por día, con la función float() puede ingresar horas fraccionarias
# y guardarlo en la variable t_spotify,t_tiktok,t_instagram,t_plataforma_streaming,t_whats
t_spotify=float(input("Ingrese el tiempo que ha escuchado Spotify en horas por día: "))
t_tiktok=float(input("Ingrese el tiempo en pantalla que paso TikTok en horas por día: "))
t_instagram=float(input("Ingrese el tiempo en pantalla que paso Instagram en horas por día: "))
t_plataforma_streaming= float(input("Ingrese el tiempo que paso en plataformas de streaming en horas por día: "))
t_whats=float(input("Ingrese el tiempo que paso en WhatsApp en horas por día: "))
#Calcular el tiempo total que paso en redes sociales y plataformas de streaming sumando las variables y guardarlo en la variable t_total.
t_total=t_spotify+t_tiktok+t_instagram+t_plataforma_streaming+t_whats
porcentaje_TD= (t_total/24) * 100
#imprimir el resultado en pantalla con la función print() y usando f-strings para mostrar el nombre de usuario, el tiempo total y el porcentaje del tiempo total del día que paso en redes sociales y plataformas de streaming que fueron reducidos a dos decimales con la funcion :.2f
print(f"El usuario {nombreusuario} acumuló un tiempo total de {t_total:.2f} horas al día en redes sociales y plataformas de streaming, lo que representa un {porcentaje_TD:.2f}% del tiempo total de su día.")
```
