## Extra 1: División de cuenta con propina
**Enunciado:** Crea un programa que pida el total de la cuenta de un restaurante,
el porcentaje de propina a dejar y el número de personas que pagarán. El programa
debe calcular el monto de la propina, el total a pagar con propina y cuánto le toca
pagar a cada persona (con dos decimales).


Modelo EPS
**Problema:** DivisionConPropina.

#### Pasos en lenguaje natural:
1. Solicitar la cantidad de la cuenta a pagar.
2. Solicitar el porcentaje de propina que desea dejar
3. Solicitar el numero de personas que van a pagar la cuenta
4. Multiplicar la cantidad de la cuenta a pagar por el pocentaje y luego dividirlo entre 100. que será nuestra propina
5. Sumar propina a la cantidad a pagar que será nuestro total
6. Dividir el total entre el numero de personas, que será el pago por persona
7. Mostrar el resultado de propina
8. Mostrar el resultaod de total con propina
9. Mostrar el resultado de pago por persona

```python
# Solicitar al usuario que ingrese el subtotal de la cuenta, el porcentaje de propina que desea dejar y el número de personas que van a pagar la cuenta. 
# Guardar estos valores en variables de tipo float (para el subtotal y el porcentaje) para realizar operaciones matemáticas con ellos.
subtotal=float(input("Ingrese la cantidad a pagar"))
porcentaje=float(input("ingrese el porcentaje de propina que desea dejar"))
# Guardar el valor en variable de tipo int (para el número de personas) para realizar operaciones matemáticas con ellos.
numero_de_personas=int(input("Ingrese el numero de personas que van a pagar la cuenta"))
#calcular la propina multiplicando el subtotal por el porcentaje y dividiendo entre 100, luego calcular el total sumando el subtotal y la propina, y finalmente calcular el total por persona dividiendo el total entre el número de personas.
propina= (subtotal*porcentaje) / 100
total= subtotal + propina
total_por_persona = total / numero_de_personas
#Imprimir el resultado en pantalla con la función print() y usando f-strings para mostrar la propina, el total con propina y el pago por persona que fueron reducidos a dos decimales con la funcion :.2f
print(f"Propina: {propina:.2f}")
print(f"Total con propina: {total:.2f}")
print(f"pago por persona: {total_por_persona:.2f}")
```

## Extra 2. Conversor de minutos a días, horas y minutos.
**Enunciado:** Crea un programa que pida una cantidad total de minutos (entero) y la convierta a días, horas y minutos restantes. Utiliza los operadores de división entera `//` y módulo `%`. (Pista: 1 día = 1440 minutos, 1 hora = 60 minutos.).


#### Pasos en lenguaje natural:
1. Solicitar la cantidad de la cantidad total de minutos
2. Calcular días dividiendo minutos entre días utilizando únicamente enteros.
3. Calcular horas utilizando el residuo de minutos entre 1440, y el residuo se divide entre 60 (valor de 1hr=60 min)
4. Calcular los minutos restantes utilizando el residuo de minutos entre 60 (valor de 1hr=60 min)
5. Mostrar el resultado de cuantos minutos equivalen en días, horas, minutos 

### PseudoCódigo (Pseint):

finAlgoritmo


```python
#solicitar al usuario que ingrese la cantidad total de minutos y guardarla en una variable de tipo int (entera) para poder realizar operaciones matemáticas con ella.
minutos=int(input("Ingrese la cantidad total de minutos: "))
#Calcular la cantidad de días, horas y minutos que equivalen a la cantidad total de minutos ingresada por el usuario usando la fórmula de conversión de minutos a días, horas y minutos.
dias=minutos//1440
horas=(minutos%1440)//60
min_rest=minutos%60
#Imprimir el resultado en pantalla con la función print() y usando f-strings para mostrar la cantidad de días, horas y minutos que equivalen a la cantidad total de minutos ingresada por el usuario.
print(f"{minutos} minuto(s) equivalen a {dias} dia(s), {horas} hora(s) y {min_rest} minuto(s)")
```

### Extra 3: Calificación final ponderada
**Enunciado:** Crea un programa que pida las calificaciones de tres parciales (valores de 0 a 10) y calcule la calificación final considerando una ponderación de **30%, 30% y 40%** respectivamente. Muestra el resultado con dos decimales.

#### Pasos en lenguaje natural:
1. Solicitar la primera calificación parcial.
2. Solicitar la segunda calificación parcial. 
3. Solicitar la tercera calificación parcial.
4. calcular el valor porcentual de la primera calificación multiplicando por .30 para simplificar la operación.
5. calcular el valor porcentual de la segunda calificación multiplicando por .30 para simplificar la operación.
6. calcular el valor porcentual de la segunda calificación multiplicando por .40 para simplificar la operación.
7. Calcular la calificación final sumando los resultados previos en el paso 4,5,6
8. Mostrar la calificacion final

### PseudoCódigo (Pseint):

Algoritmo CalcularCalificacionFinal

Escribir "Ingrese la calificación del primer parcial"

Leer parcial3

t_parcial1= parcial1 * 0.30

finAlgoritmo


```python
#Solicitar al usuario que ingrese las calificaciones de los tres parciales
#Guardarlas en variables de tipo float para poder realizar operaciones matemáticas con ellas.
parcial1=float(input("Ingrese la calificación del primer parcial(30%): "))
parcial2=float(input("Ingrese la calificación del segundo parcial(30%): "))
parcial3=float(input("Ingrese la calificación del tercer parcial(40%): "))
#Calcular el promedio de las calificaciones de los tres parciales usando la fórmula del promedio ponderado, donde el primer y segundo parcial tienen un peso del 30% y el tercer parcial tiene un peso del 40%.
t_parcial1=parcial1 * 0.30
t_parcial2=parcial2 * 0.30
t_parcial3=parcial3 * 0.40
#Calcular la calificación final sumando los resultados de los tres parciales y guardarla en una variable llamada calif_final.
calif_final= t_parcial1+t_parcial2+t_parcial3
#Imprimir el resultado en pantalla con la función print() y usando f-strings para mostrar la calificación final que fue reducida a dos decimales con la funcion :.2f
print(f"Tu calificación final es: {calif_final:.2f}")
```

#### Pasos en lenguaje natural:
1. Solicitar la cantidad en pesos mexicanos que desea convertir a otra divisa.
2. Solicitar el tipo de cambio de pesos a USD
3. Solicitar el tipo de cambio de pesos a Euros
4. Calcular el equivalente en dolares diviendo la cantidad de pesos mexicanos entre el tipo de cambio en USD
5. Calcular el equivalente en dolares diviendo la cantidad de pesos mexicanos entre el tipo de cambio en Euros
6. Mostrar la cantidad de pesos ingresada y sus equivalentes en USD y EUR

```python
#Convierte divisas, de pesos mexicanos a dólares y euros. 
# Primero se pide al usuario ingresar la cantidad en pesos mexicanos y luego el tipo de cambio a dólares y euros. 
# Al ser ingresados como datos string se convierten a float para poder realizar la operación de conversión.
cantidad_PM=int(input("Ingrese la cantidad en pesos mexicanos: "))
tipo_cambioUSD=float(input("Ingrese el tipo de cambio del dólar (USD):"))
tipo_cambioEUR=float(input("Ingrese el tipo de cambio del euro (EUR):"))
#Se realiza la conversión de pesos mexicanos a dólares y euros dividiendo la cantidad en pesos mexicanos según el tipo de cambio correspondiente.
equi_usdpm=cantidad_PM/tipo_cambioUSD
equi_eurpm=cantidad_PM/tipo_cambioEUR
# Finalmente se imprime el resultado de la conversión a dólares y euros, utilizando la función f-string y el formato .2f para limitar el número de decimales a dos.
print(f"{cantidad_PM} MXN equivalen a: ")
print(f"USD: {equi_usdpm:.2f} ")
print(f"EUR: {equi_eurpm:.2f} ")
```
