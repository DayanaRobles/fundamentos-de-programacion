## Guia para ir resolviendo el código:

### Solución.

-- CONTENIDO DEL TICKET --
- Nombre del cliente: **Listo**
- Servicios: *En proceso, falta darle formato con el ticket*
- Subtotal: **listo**
--Monto descuento y recargo---
- Monto Descuento: **Listo**
- Monto por pedido express: **listo**

- Total: **Listo**
- Tiempo estimado de entrega **Listo**

--Anticipo--
- Anticipo pagado: **Listo**
- Restante a pagar: **Listo**

---PROCESOS---
- Cantidad de pedidos a procesar (Limitarlos)
- Tipo de servicio (limitar por servicio)
- Costo prestablecidos por servicio.
- Tiempos de entrega prestablecidos según servicio: si es más de un pedido, se establece el periodo de entrega mayor.
- Descuentos: Si son mas de 2 pedidos entonces descuento, es decir apartir del 3ro descuento sobre el subtotal.
- Anticipo: Lo decide el cliente respetando que sea igual o mayor al 50%, SÍ puede dejarlo liquidado



```python
# ===COSTOS====

c_cambiosuela = 100
c_costuras = 300
c_tintado = 500
c_res_gam = 900
c_ajuste = 20 

# ===TIEMPOS DE ENTREGA EN DÍAS POR UNIDAD===

t_cambiosuela = 3
t_costuras = 5
t_tintado = 7
t_res_gam = 9
t_ajuste = 1

# Variables que cambian con el registro

subtotal = 0
t_entregafinal = 0
m_descuento = 0
m_express = 0 

# Contadores individuales para cada servicio y afinar tiempos de entrega.
cont_suela = 0
cont_costura = 0
cont_tintado = 0
cont_gamuza = 0
cont_ajuste = 0

#INICIO PROGRAMA

cliente=input("Ingrese nombre completo del cliente: ")

#Verificar que sea válida la cantidad de servicios por ticket

while True:
    c_pedidos = int(input("Ingrese cantidad de servicios a requerir por el cliente (Máximo 10): "))
    
    if 1 <= c_pedidos and c_pedidos <= 10:
        break
    else:
        print("La cantidad de servicios debe ser entre 1 y 10 por ticket.\nPor favor, intente de nuevo.")

#Despligue menu de servicios

print("\n=======MENU========")
print("========DE=========")
print("=====SERVICIOS=====\n")
print(" (1)Cambio Suela\n (2)Costuras\n (3)Tintado\n (4)Restauración gamuza\n (5)Ajuste de correa")

#Inicia resgistro de pedidos.
print("\n=======================")
print("=========Ticket========")
print("=======================")
print(f"Nombre cliente: {cliente}")


for pedido in range (1, c_pedidos + 1):

    #Verificar que exista la opición seleccionada: 
    #si existe entonces verificar si ya llegó al limite por servicios solicitados. 
    
    while True:
        
        tiposervicio=int(input(f"Ingrese número de servicio no.{pedido}: "))
        
        if tiposervicio >= 1 and tiposervicio <= 5:
          
            if tiposervicio == 1 and cont_suela >= 3:
                print("Se llegó al límite de cantidad de servicios posibles en -Cambio de Suela-.\nElija otro servicio.")
        
            elif tiposervicio == 2 and cont_costura >= 3:
                print("Se llegó al límite de cantidad de servicios posibles en -Costuras-.\nElija otro servicio.")
        
            elif tiposervicio == 3 and cont_tintado >= 2:
                print("Se llegó al límite de cantidad de servicios posibles en -Tintado-.\nElija otro servicio.")
        
            elif tiposervicio == 4 and cont_gamuza >= 2:
                print("Se llegó al límite de cantidad de servicios posibles en -Restauración de Gamuza-.\nElija otro servicio.")
        
            elif tiposervicio == 5 and cont_ajuste >= 9:
                print("Se llegó al límite de cantidad de servicios posibles en -Ajuste de correas-.\nElija otro servicio.")
                
            else: 
                break
        else:
            print("Opción no válida.\nPor favor, Ingrese opción del 1 al 5")              

    #Procesos internos dependiendo del servicio: 

    if tiposervicio == 1:
        print(f"Costo servicio 'Cambio de suela': ${c_cambiosuela:.2f}")
        subtotal += c_cambiosuela
        cont_suela += 1

    #Regla: Tiempo base por un par es 3 días, por cada par extra 2 días más

        t_servicio = t_cambiosuela + (cont_suela - 1) * 2
        if t_servicio > t_entregafinal:
            t_entregafinal = t_servicio

    elif tiposervicio == 2:
        print(f"Costo servicio 'Costura': ${c_costuras:.2f}")
        subtotal += c_costuras
        cont_costura += 1 

    #Regla: Tiempo base por un par es 5 días, por cada par extra 3 días más
        
        t_servicio = t_costuras + (cont_costura - 1) * 3
        if t_servicio > t_entregafinal:
            t_entregafinal = t_servicio    

    elif tiposervicio == 3:
        print(f"Costo servicio 'Tintado': ${c_tintado:.2f}")
        subtotal += c_tintado
        cont_tintado  += 1

     #Regla: Tiempo base por un par es 7 días, por cada par extra 4 días más
        
        t_servicio = t_tintado + (cont_tintado - 1) * 4
        if t_servicio > t_entregafinal:
            t_entregafinal = t_servicio

    elif tiposervicio == 4:
        print(f"Costo servicio 'Restauración Gamuza': ${c_res_gam:.2f}")
        subtotal += c_res_gam
        cont_gamuza += 1

    #Regla: Tiempo base por un par es 9 días, por cada par extra 5 días más
                
        t_servicio = t_res_gam + (cont_gamuza - 1) * 5
        if t_servicio > t_entregafinal:
            t_entregafinal = t_servicio

    
    elif tiposervicio == 5:
        print(f"Costo servicio 'Ajuste Correa': ${c_ajuste:.2f}")
        subtotal += c_ajuste
        cont_ajuste += 1
       
        #Mismo día
        if cont_ajuste <= 2:
            t_servicio= 0  
        #Día siguiente
        elif cont_ajuste <= 5:
            t_servicio = t_ajuste 
        # A partir de 6 pares: Tiempo base 1 día, por cada par extra 1 día más.
        else:
            t_servicio = t_ajuste + (cont_ajuste - 5) * 1
            
        if t_servicio > t_entregafinal:
            t_entregafinal = t_servicio
    
    
#Termina registro servicios.
print(f"subtotal= ${subtotal:.2f}")
#Muestra si aplica descuento (10%) y si no el monto de descuento es 0

if c_pedidos > 2:
    print("Servicio con descuento (10%)")
    m_descuento = subtotal * 0.10
else:
    m_descuento=0    

#Pregunta si el pedido es urgente
#solo aplica cuando el tiempo de entrega es mayor a un día

if t_entregafinal > 1: 
    while True:
        urgente=input("\n¿El pedido es urgente?\n si/no :")
        #Verificar que sea válida la opcion
        if urgente == "si" or urgente == "Si" or urgente == "SI" or urgente == "no" or urgente == "No" or urgente == "NO":
            break
        print("Opción no válida. Por favor responda 'si' o 'no'.")

    #si es urgente aplicar el 20% de recargo y reducir el tiempo de entrega
    #Únicamente días enteros.
    
    if urgente == "si" or urgente == "Si" or urgente == "SI":
        m_express = subtotal * 0.20  
        t_entregafinal = t_entregafinal // 2
        print("Servicio Express (+20%) ")
    else:
        m_express=0
#Si la entrega es para el mismo día, en caso de que sea ajuste de correa si es unitaria puede ser el mismo día cuando se entregue.
else:
    urgente= "no"
    m_express = 0
    print("Este pedido N/A para servicio Express. Tiene un tiempo de entrega de un día habil")
    
#Operación total: se aplican descuentos y recargos.
total = subtotal - m_descuento + m_express



#Mostrar sí hay descuento o Recargos, y sus respectivos montos

if m_descuento > 0:
    print(f"Monto descuento: ${m_descuento:.2f}")
    
if m_express > 0:
    print(f"Monto Servicio Express: ${m_express:.2f}")

#Mostrar total

print (f"TOTAL= ${total:.2f}")

#Tiempos de entrega

if t_entregafinal == 0:
    print("Tiempo estimado de entrega: Mismo día")
else: 
    print("Tiempo estimado de entrega: ",t_entregafinal, "días, después de anticipo")

#Anticipos: Puede ser ingresado por el cliente
#Reglas: Minimo 50% y Máximo el TOTAL del pedido

#Operación anticipo
anti_min = total * 0.50

#El cliente puede decidir el anticipo: debe ser mínimo el 50% o dejar liquidado.
while True:
    anticipo = float(input(f"Ingrese el anticipo que desea dejar el cliente (Mínimo 50%, equivale a ${anti_min}.\nAnticipo cliente: $"))

    if anticipo >= anti_min and anticipo <= total:
        break
    elif anticipo > total:
        print(f"El anticipo no puede ser mayor al total: ${total:.2f}")
    else: 
        print(f"El anticipo debe ser al menos del 50%: ${anti_min:.2f}")

#Calcular el saldo pediente
s_pendiente= total-anticipo

#Mostrar cantidad del anticipo y restante a pagar.
print(f"Anticipo pagado = ${anticipo:.2f}")

if anticipo == total:
     print(f"Producto(s) liquidado")

else:
    print(f"Restante a pagar= ${s_pendiente:.2f}")

print("\n========¡Buen día!========\nGracias por elegir 'RENOVACALZADO'")
    
```
