### Solución.

-- CONTENIDO DEL TICKET --
- Nombre del cliente: **Listo**
- Servicios: *En proceso, falta darle formato con el ticket*
- Subtotal: **listo**
--Monto descuento y recargo---
- Monto Descuento: **Listo**
- Monto por pedido express: **listo**

- Total: **Listo**
- Tiempo estimado de entrega *Por resolver*

--Anticipo--
- Anticipo pagado: **Listo**
- Restante a pagar: **Listo**

---PROCESOS---
- Cantidad de pedidos a procesar.
- Tipo de servicio
- Costo prestablecidos por servicio.
- Tiempos de entrega prestablecidos según servicio: si es más de un pedido, se establece el periodo de entrega mayor.
- Descuentos: Si son mas de 2 pedidos entonces descuento, es decir apartir del 3ro descuento sobre el subtotal.
- Anticipo: Lo decide el cliente respetando que sea igual o mayor al 50%, SÍ puede dejarlo liquidado



```python
#===COSTOS====
c_cambiosuela = 100
c_costuras = 300
c_tintado = 500
c_res_gam = 900
c_tala = 20 
#===TIEMPOS DE ENTREGA===
t_cambiosuela = 3
t_costuras = 5
t_tintado = 7
t_res_gam = 9
t_tala = 1
#Variables que cambian con el registro

subtotal=0
tiempo_f=0

##INICIO PROGRAMA

cliente=input("Ingrese nombre del cliente: ")
c_pedidos=int(input("Ingrese cantidad de servicios a requerir por el cliente: "))

#Ticket servicio costo individual.

# ya no sirve este while c_pedidos >= registro:
print("=======MENU========")
print("========DE=========")
print("=====SERVICIOS=====")
print(" (1)Cambio Suela\n (2)Costuras\n (3)Tintado\n (4)Restauración gamuza\n (5)Talabartería")


for pedido in range (1, c_pedidos + 1):
    ##Garantizar que sí exista la opción.
    while True:
        tiposervicio=int(input(f"Ingrese número de servicio no.{pedido}: "))

        if tiposervicio>=1 and tiposervicio<=5:
            break
        else:
            print("Opción no válida.\n Por favor, Ingrese opción del 1 al 5")              
    if tiposervicio == 1:
        print(f"Costo servicio cambio de suela/tacón: {c_cambiosuela}")
        subtotal +=c_cambiosuela
        if tiempo_f < t_cambiosuela:
            tiempo_f=t_cambiosuela
    elif tiposervicio == 2:
        print(f"Costo servicio Costura: {c_costuras}")
        subtotal +=c_costuras
        if tiempo_f < t_costuras:
            tiempo_f=t_costuras
    elif tiposervicio == 3:
        print(f"Costo servicio Tintado: {c_tintado}")
        subtotal +=c_tintado
        if tiempo_f < t_tintado:
            tiempo_f=t_tintado
    elif tiposervicio == 4:
        print(f"Costo servicio Restauración Gamuza: {c_res_gam}")
        subtotal += c_res_gam
        if tiempo_f < t_res_gam:
             tiempo_f=t_res_gam
    elif tiposervicio == 5:
        print(f"Costo servicio Tala: {c_tala}")
        subtotal +=c_tala
        if tiempo_f < t_tala:
            tiempo_f=t_tala

#Termina registro servicios.

#MOSTRAR Y HACER DESCUENTOS SI APLICA
if c_pedidos > 2:
    print("Aplica descuento")
    montodes = subtotal * 0.10
else:
    montodes=0    

# Recargo por pedido express si aplica.
#FALTA VALIDACIÓN...
urgente=input("\n¿El pedido es urgente?\n si/no :")

if urgente == "si":
    montoUrg = subtotal * 0.20  
    print(" Servicio Seleccionado como Express (+20%) ")
else:
    montoUrg=0

#Operacion total

total = subtotal - montodes + montoUrg

#Resumen pedido

print("\n=======================")
print("=========Ticket========")
print("=======================")
print(f"Cliente: {cliente}")

print("Tiempo estimado de entrega: ",tiempo_f, "días, después de anticipo")
print(f"subtotal= ${subtotal}")

#Variaciones sí hay descuento o Recargos

if montodes > 0:
    print(f"Descuento (10% por volumen): {montodes:.2f}")
    
if urgente == "si":
    print(f"Monto pedido express: {montoUrg:.2f}")

#Mostrar total finañ

print (f"TOTAL= ${total:.2f}")

#Anticipos: Puede ser ingresado por el cliente
#Reglas: Minimo 50% y Máximo el TOTAL del pedido

anti_min = total* 0.50

#Validar
while True:
    anticipo = float(input(f"Ingrese el anticipo del cliente (Mínimo 50%, equivale a ${anti_min}.\nAnticipo cliente: $"))

    if anticipo >= anti_min and anticipo <= total:
        break
    elif anticipo > total:
        print(f"El anticipo no puede ser mayor al total: {total}")
    else: 
        print(f"El anticipo debe ser al menos del 50%: {anti_min}")

s_pediente= total-anticipo

#Mostrar valores de anticipo y restante a pagar.
print(f"Anticipo pagado = {anticipo:.2f}")

if anticipo == total:
     print(f"Dejo liquidado")

else:
    print(f"Restante a pagar al entregar= {s_pediente:.2f}")

    
```
