## Extra 1: Suma de todos los elementos de una matriz
**Enunciado:** Crea una función `suma_matriz(matriz)` que **regrese valor** (con `return`) la suma de todos los elementos de una lista de listas, usando ciclos `for` anidados. Define además una función `imprimir_matriz(matriz)` que **no regrese valor** y muestre la matriz sin corchetes ni comas. Aplica ambas funciones a la matriz `[[1, 2, 3], [4, 5, 6], [7, 8, 9]]`.
**Entrada:**
```
Matriz: [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
```
**Salida:**
```
1 2 3
4 5 6
7 8 9
La suma de todos los elementos de la matriz es: 45
```

### Planteamiento lógico aplicado:

- Repetición: se utilizan ciclos for anidados.

- Estructura: se utiliza lista de listas para formar una matriz de 3 x 3

- Retorno de valor: se utiliza return para devolver el resultado de la suma

1. Se inicializa la matriz con los datos.

2. La función imprimir_matriz sirve para mostrar la matriz. Con dos for, recorre las filas y las columnas, dejando un espacio entre los números (columnas) con end=" " y con print() se cambia de fila

3. La función suma_matriz inicializa un acumulador(suma) en 0. Después recorre todos los elementos de la matriz y los va sumando en esa variable.

4. Finalmente, se llama a suma_matriz y el resultado se guarda en resultado, para después mostrar la suma de todos los elementos de la matriz.

````python
# Función que calcula la suma total.
def suma_matriz(matriz):
    suma = 0
    #Primer ciclo recorre fila por fila
    for fila in matriz:
        #Segundo ciclo recorre la fila actual y toma cada número
        for elemento in fila:
            suma += elemento
    # Regresa el resultado final
    return suma
# Función que muestra la matriz sin comas ni corchetes
# NO regresa valor
def imprimir_matriz(matriz):
    
    for fila in matriz:
        for columna in fila:
            print(columna, end=" ")
        print()


matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
#Manda a llamar la función e imprime.
imprimir_matriz(matriz)
#Se guarda en resultado el valor que regresó la función
resultado = suma_matriz(matriz)
print(f"La suma de todos los elementos de la matriz es: {resultado}")
````

### Extra 2: Matriz con suma por fila
**Enunciado:** Crea una función `suma_fila(fila)` que **regrese valor** (con `return`) la suma de los elementos de una lista, y una función `mostrar_tabla(matriz)` que **no regrese valor** e imprima la matriz sin corchetes ni comas. Dada la matriz `[[3, 1, 4], [1, 5, 9], [2, 6, 5]]`, imprime la matriz y la suma de cada fila.
**Entrada:**
```
Matriz: [[3, 1, 4], [1, 5, 9], [2, 6, 5]]
```
**Salida:**
```
3 1 4
1 5 9
2 6 5
Suma de la fila 0: 8
Suma de la fila 1: 15
Suma de la fila 2: 13
```

El orden de la matriz siempre es: Matriz[FILA][COLUMNA]

### Planteamiento lógico aplicado:

- Repetición: se utilizan ciclos for anidados.

- Estructura: se utiliza lista de listas para formar una matriz de 3 x 3

- Retorno de valor: se utiliza return para devolver el resultado de la suma de cada fila.


1. Se inicializa la matriz con los datos.

2. La función mostrar_tabla sirve para mostrar la matriz. Con dos for, recorre las filas y las columnas, dejando un espacio entre los números (columnas) con end=" " y con print() se cambia de fila

3. La función suma_fila inicializa un acumulador(suma) en 0. Después recorre todos los elementos de la fila y los va sumando en esa variable.

4. Finalmente, se utiliza un ciclo for con range para recorrer cada fila de la matriz. Se llama a suma_fila, para obtener la suma de cada fila

````python
# Función que calcula la suma total de los elementos del número de fila que se ingreso
def suma_fila(fila):
    suma=0
    for columna in fila:
        suma += columna
    return suma

# Función que muestra la matriz sin comas ni corchetes
# NO regresa valor
def mostrar_tabla(matriz):
    for fila in matriz:
        for columna in fila:
            print(columna, end=" ")
        print()

Matriz = [[3, 1, 4], [1, 5, 9], [2, 6, 5]]
mostrar_tabla(Matriz)

# Recorre cada fila dentro del rango 0 a 2 para calcular e imprimir la suma.
for i in range(len(Matriz)):
    resultado = suma_fila(Matriz[i])
    print(f"Suma de la fila {i}: {resultado}")
````

### Extra 3: Producto de una columna sin usar el operador *
**Enunciado:** Crea la matriz `[[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]]` (4 filas y 3 columnas). Define una función `multiplicar_columna(matriz, columna)` que **regrese valor** (con `return`) el producto de los elementos de una columna dada. **Restricción:** dentro de la función está prohibido usar el operador `*`; el producto debe calcularse con sumas repetidas usando un ciclo. Pide al usuario el número de columna (`0`, `1` o `2`) y muestra el resultado.
**Entrada:**
```
Numero de columna: 1
```
**Salida:**
```
El producto de la columna 1 es: 320
```
*(Los elementos de la columna 1 son `2`, `5`, `8` y `4`; su producto es `2 * 5 * 8 * 4 = 320`.)*

### Planteamiento lógico aplicado:

- Repetición: se utilizan ciclos for anidados, uno para guardar los elementos de la columna y otro para hacer las sumas repetidas.

- Estructura: se utiliza lista de listas para formar una matriz de 4 x 3 y una lista simple para guardar los números de la columna (elementos)

- Retorno de valor: se utiliza return para devolver el resultado de la multiplicación de los elementos de la columna

- Validación: se utiliza un ciclo while para verificar que sea una columna válida.


1. Se inicializa la matriz con los datos.

2. La función multiplicar_columna extrae el numero de la columna pedida de cada fila y crea una lista con ellos (elementos)

3. Toma el primer número como resultado base, y con dos for, va sumando ese valor las veces que indique el siguiente número.

4. Finalmente, se utiliza un while para validar que este entre 0 y 2 y entonces mandar a llamar a la función. Se muestra el resultado

````python
matriz=[
    [1, 2, 3], 
    [4, 5, 6], 
    [7, 8, 9], 
    [2, 4, 6] 
    ]
# Función que extrae la columna y multiplica sin usar *
def multiplicar_columna(matriz,columna):
    elementos=[]
    # Recorre las filas y guarda solo el número de las columna elegida
    for lista in matriz:
        factor = lista[columna]
        elementos.append(factor)
    
    resultado = elementos[0]
    #  Recorre los demás elementos y los multiplica con sumas repetidas
    for i in range (len(elementos)-1):
        base = resultado  
        factor_sig=elementos[i + 1]
        for v in range(factor_sig -1):
            resultado +=base

    #Regresa el resultado de la multiplicación
    return resultado

#Valida que la columna este en un rango de 0-2 
while True:
    c=int(input("Ingrese columna: "))

    if c >=0 and c<=2:
        break
    else:
        print("Error: La columna debe ser 0, 1 o 2.")

#Llama la función e imprime resultado
print(f"El producto de la columna {c} es: {multiplicar_columna(matriz,c)}")
        
````

### Extra 4: Tabla de Pitágoras de tamaño variable
**Enunciado:** Crea una función `generar_tabla(tamano)` que **regrese valor** (con `return`) una lista de listas de `tamano x tamano` con los productos de la tabla de Pitágoras (renglones y columnas del 1 al `tamano`). Dentro de la función de generación está prohibido usar el operador `*`; usa sumas repetidas con ciclos. Define una función `imprimir_tabla(tabla)` que **no regrese valor** e imprima la tabla sin corchetes ni comas, con los números tabulados (`\t`). Pide al usuario un tamaño entre 2 y 5; si ingresa un valor fuera del rango, muestra un mensaje de error. Al final pide un renglón y una columna y muestra el producto consultando la matriz, sin usar `*`.
**Entrada:**
```
Tamano de la tabla: 4
Renglon (factor): 2
Columna (factor): 3
```

### Planteamiento lógico aplicado:

- Repetición: se utilizan ciclos for anidados, para construir la tabla con sumas repetidas e imprimir los valores con tabuladores.

- Estructura: se utiliza lista de listas para formar la tabla de Pitágoras según el tamaño que ingrese el usuario (n x n)

- Retorno de valor: generar_tabla regresa la matriz construida y obtener_multiplicación regresa el valor consultado

- Validación: se utiliza un ciclo while para verificar que el tamaño esté entre 2 y 5 y que los factores ingresados no salgan del rango


1. Se pide al usuario el tamaño de la tabla mediante un while que valide que esté entre 2 y 5

2. La función generar_tabla construye la matriz multiplicando mediante un for que realiza sumas repetidas, y regresa la matriz con return

3. La función imprimir_tabla, muestra los encabezados superiores y laterales juntos con los valores previamente obtenidos en generar_tabla usando \t para los espacios.

4. Finalmente, se solicitan Factor 1 y Factor 2 validando que estos esten dentro del rango permitido, de ser así, mediante la función se  obtiene el resultado final y se muestra.

````python
# Función que genera la tabla, usando solo sumas repetidas
def generar_tabla(tamano):
    n = tamano
    tabla=[]

    for f in range(1, n + 1):
        fila_actual=[]
        for c in range(1,n + 1):
            valor= 0
            # Suma f las veces que indique c 
            for v in range (c):
                valor +=f
            fila_actual.append(valor)   
        tabla.append(fila_actual)

    return tabla

# Función que muestra la matriz sin comas ni corchetes
# NO regresa valor
def imprimir_tabla(matriz):
    # Espacio esquina superior izquierda que sirve para alinear 
    print (" ", end="\t")
    # Encabezado horizontal 
    for c in range(1,len(matriz)+1):
        print(c, end="\t")
    print()
    # Encabezado lateral y resultados de la multiplicación
    for f in range(1, len(matriz)+1):
        print(f, end="\t")
        for c in range(1,len(matriz)+1):
            print(matriz[f-1][c-1], end="\t")
        print()
# Función que consulta el resultado
def obtener_multiplicación(matriz, factor1, factor2):
     return matriz[factor1-1][factor2-1]
#Valida que el tamaño ingresado este en un rango de 2 a 5
while True:
    tamano=int(input("Ingrese el tamaño: "))
    if tamano>=2 and tamano<=5:
        break
    else:
        print("Error: El tamaño debe estar entre 2 y 5.")
    
#Genera e imprime la tabla
tabla= generar_tabla(tamano)
imprimir_tabla(tabla)
#Valida que factor 1 esté en un rango del 1 al tamaño definido por el usuario
#Repite hasta recibir un valor entre ese rango
while True:
    f1 = int(input("Ingrese Factor 1: "))
    if f1 >=1 and f1<=tamano:
        break
    else: 
        print(f"Error: El número debe estar entre 1 y {tamano}.\nPor favor, elija un número en el rango")

#Valida que factor 2 esté en un rango del 1 al tamaño definido por el usuario
#Repite hasta recibir un valor entre ese rango
while True:
   f2 = int(input("Ingrese Factor 2: "))
   if f2>0 and f2<=tamano:
        break
   else: 
        print(f"Error: El número debe estar entre 1 y {tamano}.\nPor favor, elija un número en el rango")
   

resultado= obtener_multiplicación(tabla, f1, f2)
print(f"El producto de {f1} x {f2} : {resultado}")
````

3. Se muestra la lista completa con los datos ingresados.

4. Con un ciclo for, se recorre cada número de la lista: si es negativo, se calculan e imprimen los espacios necesarios y los asteriscos a la izquierda de la barra |; si es positivo o cero, se imprimen 10 espacios, la barra | y los asteriscos a la derecha.

````python
lista = []
# Pide los 10 números, todos en un rango entre -10 y 10
for i in range(10):
    while True:
         numero = int(input(f"Ingresa un numero del -10 al 10: "))
         if -10 <= numero <= 10:
             lista.append(numero)
             break
         else:
             print("Por favor, ingresa un número dentro del rango permitido.")

print("Lista completa:", lista)
print("Impresión de asteriscos según los números ingresados:")
# Recorre los numeros guardados para graficar
for i in lista:
    if i < 0:
        # Calcula los espacios mediante una resta
        espacios=10-(-i)
        # Crea los espacios
        for esp in range(espacios):
            print(" ", end="")
        # Dibuja los asteriscos
        for j in range(-i):
            print("*", end="")
        # Coloca la barra central    
        print("|", end="")
        
    else:
        espacios=10
        # Coloca la barra central
        for esp in range(espacios):
            print(" ", end="")
        print("|", end="")
        # Dibuja los asteriscos a la derecha 
        for j in range(i):
            print("*", end="")
    # Salto de línea para el siguiente número a graficar
    print()  
````
