```python
#Longitud
n=10
tabla=[]

for f in range(0,n+1):
    fila_actual=[]
    for c in range(0,n+1):
        #Espacio en blanco esquina    
        if f==0 and c==0:
            fila_actual.append(" ")
        #Fila 0: encabezado
        elif f == 0:
            fila_actual.append(c)
        #Columna 0: Laterales
        elif c==0:
            fila_actual.append(f)
        #llenar tabla de pitagoras
        else:
            valor=f * c
            fila_actual.append(valor)   
            
    tabla.append(fila_actual)

#Función imprimir matriz
def mostrar_tabla(matriz):
    for f in range(0,n + 1):
        for c in range(0, n +1):
            print(matriz[f][c], end="\t")
        print()
#Como los encabezados ocupan la fila 0 y la columna 0, los numeros a multiplicar
#coinciden con los índices de la matriz sin necesidad de restar 1; a diferencia de la opción sugerida por el profesor
def obtener_multiplicación(matriz, factor1, factor2):
    
   return matriz[factor1][factor2]
    
mostrar_tabla(tabla)

f1 = int(input("Ingrese Factor1: "))
f2 = int(input("Ingrese Factor2: "))

resultado= obtener_multiplicación(tabla, f1, f2)
print("============================================================================================")
print("============================================================================================")
print("============================================================================================")
print(f"El resultado de {f1} x {f2} : {resultado}")
```
