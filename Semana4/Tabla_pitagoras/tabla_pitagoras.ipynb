Desarrollar una **aplicación en Python** que muestre una **tabla de Pitágoras interactiva** para facilitar la multiplicación de dos factores ingresados por el usuario. El programa debe almacenar la tabla en una **lista de listas (matriz)**, imprimirla en pantalla de forma ordenada y permitir que el usuario capture dos factores (renglón y columna) para obtener su producto consultando directamente la información almacenada en la matriz.

### Planteamiento lógico aplicado:

- Repetición: se utilizan ciclos for anidados para crear la matriz y despues mostrar sus valores usando tabulador

- Decisión: Se utiliza if/elif/else para crear los encabezados lateral y superior (fila 0 y columna 0), y el producto en el interior de la tabla

- Estructura: se utiliza una lista de listas para formar la tabla de Pitágoras y guardar los datos.


- Validación: se usa un ciclo while para comprobar que los factores estén entre 1 y 10.


1. Se define la dimensión n=19 y se inicializa una matriz vacia (tabla)

2. Con dos ciclos anidados,Con if/elif/else se coloca un espacio en [0][0], los encabezados en la primera fila y columna, y en las demás celdas se guarda el resultado de f * c.

3. La función mostrar_tabla recorre e imprime la matriz completa usando \t para alinear horizontalmente todas las columnas.

4. La función obtener_multiplicación busca directamente el resultado en matriz[factor1][factor2], ya que los índices corresponden con los números de los encabezados.

5. Se piden el Factor 1 y el Factor 2 mediante ciclos while, comprobando que los valores estén entre 1 y 10. Después se llama a la función y se imprime el resultado.

```python
# Longitud de la tabla
n = 10

tabla=[]

#Construcción de la matriz: 
#Recorre cada fila de 0 a n (10)
#Decidí hacerla así para ocupar mi fila 0 como encabezado.
#Y mi columna 0 como mi encabezado vertical

for f in range(0,n+1):
    fila_actual=[]
    #Recorre cada columna de la fila actual.
    for c in range(0,n+1):

        #Espacio en blanco esquina superior izquierda para alinear encabezados    
        if f==0 and c==0:
            fila_actual.append(" ")

        #Fila 0: genera el encabezado horizontal
        elif f == 0:
            fila_actual.append(c)

        #Columna 0: genera el encabezado lateral
        elif c==0:
            fila_actual.append(f)

        # Columnas internas:
        #calcula y guarda la multiplicación de fila por columna actual
        else:
            valor=f * c
            fila_actual.append(valor)   

    #Guarda la fila completa en la tabla.       
    tabla.append(fila_actual)

#Función que imprime la matriz, sin comas ni corchetes.
def mostrar_tabla(matriz):
    for f in range(len(matriz)):
        for c in range(len(matriz)):
            #end="\t" es un tabulador que nos ayuda a dar formato visual de cuadrícula
            print(matriz[f][c], end="\t")
        print()

#Nota explicativa:
#Como los encabezados ocupan la fila 0 y la columna 0,
#Los números a multiplicar coinciden con los índices 
# de la matriz sin necesidad de restar 1;
# a diferencia de la opción sugerida en el documento por el profesor
def obtener_multiplicación(matriz, factor1, factor2):
    
   return matriz[factor1][factor2]
    
mostrar_tabla(tabla)

#Valida que factor 1 esté en un rango del 1 al 10
#Repite hasta recibir un valor entre ese rango
while True:
    f1 = int(input("Ingrese Factor 1: "))
    if f1>0 and f1<=n:
        break
    else: 
        print("Error: El número debe estar entre 1 y 10.\nPor favor, elija un número en el rango")

#Valida que factor 2 esté en un rango del 1 al 10
#Repite hasta recibir un valor entre ese rango
while True:
   f2 = int(input("Ingrese Factor 2: "))
   if f2>0 and f2<=n:
        break
   else: 
        print("Error: El número debe estar entre 1 y 10.\nPor favor, elija un número en el rango")
   

resultado= obtener_multiplicación(tabla, f1, f2)
print("============================================================================================")
print("============================================================================================")
print("============================================================================================")
print(f"El resultado de {f1} x {f2} es: {resultado}")
```
